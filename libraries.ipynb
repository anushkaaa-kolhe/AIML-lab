```python
import math

print("Value of Pi:", math.pi)
print("Square root of 25:", math.sqrt(25))
print("2 raised to power 4:", math.pow(2, 4))
print("Factorial of 6:", math.factorial(6))
print("Sine of 90 degrees:", math.sin(math.radians(90)))
print("Cosine of 0 degrees:", math.cos(math.radians(0)))
print("Natural logarithm of 10:", math.log(10))
print("Ceiling of 4.3:", math.ceil(4.3))
print("Floor of 4.8:", math.floor(4.8))
```

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

print("Array:", data)
print("Sum:", np.sum(data))
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", np.std(data))
print("Maximum:", np.max(data))
print("Minimum:", np.min(data))

matrix = np.array([[1, 2], [3, 4]])

print("\nMatrix:")
print(matrix)

print("\nTranspose:")
print(matrix.T)

print("\nMatrix Multiplication:")
print(np.dot(matrix, matrix))
```

```python
import numpy as np
from scipy import stats

data = np.array([20, 22, 25, 28, 30, 32, 35])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", np.std(data))
print("Variance:", np.var(data))

mode_result = stats.mode(data, keepdims=True)
print("Mode:", mode_result.mode[0])
```

```python
import pandas as pd

data = {
    "Name": ["Ankita", "Ravi", "Nishi", "Sneha", "Anu"],
    "Marks": [85, 78, 92, 88, 95],
    "Age": [20, 21, 20, 22, 20]
}

df = pd.DataFrame(data)

print("DataFrame:")
print(df)

print("\nFirst 3 Records:")
print(df.head(3))

print("\nAverage Marks:", df["Marks"].mean())

print("\nMaximum Marks:", df["Marks"].max())

print("\nStudents with Marks Greater Than 85:")
print(df[df["Marks"] > 85])

print("\nStatistical Summary:")
print(df.describe())
```

```python
import matplotlib.pyplot as plt

names = ["Ankita", "Ravi", "Nishi", "Sneha", "Anu"]
marks = [85, 78, 92, 88, 95]

plt.bar(names, marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Name": ["Ankita", "Ravi", "Nishi", "Sneha", "Anu"],
    "Marks": [85, 78, 92, 88, 95]
}

df = pd.DataFrame(data)

sns.barplot(x="Name", y="Marks", data=df)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 60, 62, 65, 68, 70, 72, 75, 80, 85, 90, 95]

sns.histplot(marks, bins=5, kde=True)

plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Marks")
plt.show()
```
