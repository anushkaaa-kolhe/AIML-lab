```python
board = [" "] * 9

def display():
    print()
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print()

def check_winner(player):
    wins = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]

    return any(board[a] == board[b] == board[c] == player
               for a, b, c in wins)

def minimax(is_ai):
    if check_winner("O"):
        return 1
    if check_winner("X"):
        return -1
    if " " not in board:
        return 0

    if is_ai:
        best = -float("inf")
        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                score = minimax(False)
                board[i] = " "
                best = max(best, score)
        return best

    best = float("inf")
    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            score = minimax(True)
            board[i] = " "
            best = min(best, score)
    return best

def ai_move():
    best_score = -float("inf")
    move = 0

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            score = minimax(False)
            board[i] = " "

            if score > best_score:
                best_score = score
                move = i

    board[move] = "O"

def game():
    print("Tic-Tac-Toe: You are X, AI is O")

    for turn in range(9):
        display()

        if turn % 2 == 0:
            while True:
                try:
                    position = int(input("Enter position (1-9): ")) - 1

                    if position not in range(9):
                        print("Enter a number between 1 and 9.")
                    elif board[position] != " ":
                        print("Position already occupied.")
                    else:
                        board[position] = "X"
                        break
                except ValueError:
                    print("Enter a valid number.")

            if check_winner("X"):
                display()
                print("You win!")
                return

        else:
            print("AI is making a move...")
            ai_move()

            if check_winner("O"):
                display()
                print("AI wins!")
                return

    display()
    print("It's a draw!")

game()6
```
